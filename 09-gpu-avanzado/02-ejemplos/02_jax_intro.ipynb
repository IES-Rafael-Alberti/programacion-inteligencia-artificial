# Introducción a JAX — Base


JAX ofrece **arrays como NumPy**, compilación JIT y autodiferenciación sobre distintos backends.

**Objetivos**
- Crear arrays y operar con `jax.numpy`.
- Usar `jax.grad` para derivar funciones.
- Identificar el backend y el dispositivo real de ejecución.

Este ejemplo puede ejecutarse en CPU para aprender o depurar. Una salida CPU **no sirve como evidencia evaluable de aceleración GPU** en UD9.

```
try:
    import jax
    import jax.numpy as jnp

    HAS_JAX = True
    JAX_DEVICES = jax.devices()
    JAX_GPU_AVAILABLE = any(device.platform == "gpu" for device in JAX_DEVICES)
    print("✅ JAX", jax.__version__)
    print("Backend predeterminado:", jax.default_backend())
    print("Dispositivos:", JAX_DEVICES)
    if not JAX_GPU_AVAILABLE:
        print("⚠️ Modo CPU: válido para demostración/depuración, NO como evidencia GPU evaluable.")
except Exception as e:
    HAS_JAX = False
    JAX_GPU_AVAILABLE = False
    print("⚠️ JAX no disponible:", e)
```

```
if HAS_JAX:
    x = jnp.linspace(-3, 3, 7)
    f = lambda t: jnp.sin(t) * jnp.exp(-t**2)
    g = jax.grad(lambda t: jnp.sum(f(t)))
    print('x =', x)
    print('f(x) =', f(x))
    print('∂/∂x sum f(x) evaluado en x[0]:', float(g(x)[0]))
else:
    print('Instala jax + jaxlib para probar.')
```
