```

import numpy as np
from pathlib import Path

# Cargar modelo si existe (de 102); si no, comportamiento dummy
scaler = model = None
try:
    import joblib
    if Path("artifacts/model.joblib").exists():
        bundle = joblib.load("artifacts/model.joblib")
        scaler, model = bundle["scaler"], bundle["model"]
except Exception:
    pass

def predictor(f1: float, f2: float) -> float:
    x = np.array([[f1, f2]], dtype="float32")
    if scaler is not None and model is not None:
        xs = scaler.transform(x)
        return float(model.predict(xs)[0])
    return float(1.0 if (f1+f2)>0 else 0.0)

out_a = predictor(1.0, 1.0)
out_b = predictor(-2.0, -2.0)
```

```

# Helper de autotests
def check(cond, ok_msg, fail_msg):
    if cond:
        print("✔", ok_msg)
    else:
        raise AssertionError("✘ " + fail_msg)
```

```

# Tests
check(out_a in [0.0,1.0], "Predicción válida para caso A.", "Salida A fuera de {0,1}.")
check(out_b in [0.0,1.0], "Predicción válida para caso B.", "Salida B fuera de {0,1}.")
```
