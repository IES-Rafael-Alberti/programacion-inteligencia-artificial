# 13 - Plotly interactivo — **Soluciones + Autotests**


Ejemplos con **plotly.express**: scatter, facetas, barras, histogramas, box y exportación a HTML.


## 1) Dataset y gráficos base

```

import numpy as np, pandas as pd
rng = np.random.default_rng(0)
n = 400
df = pd.DataFrame({
    "x": rng.normal(0,1,n),
    "y": rng.normal(0,1,n) + np.linspace(-1,1,n),
    "cat": rng.choice(["A","B","C"], n),
    "tam": rng.integers(5,20,n)
})
df.shape
```

```

import plotly.express as px

fig_scatter = px.scatter(df, x="x", y="y", color="cat", size="tam", hover_data=["tam"])
fig_scatter
```

## 2) Facetas y subplots

```

fig_facet = px.scatter(df, x="x", y="y", color="cat", facet_col="cat", trendline="ols")
fig_facet
```

## 3) Barras, histogramas y boxplots

```

fig_bar = px.bar(df.groupby("cat")["y"].mean().reset_index(),
                 x="cat", y="y", title="Media de y por categoría")
fig_hist = px.histogram(df, x="x", nbins=30, marginal="box", title="Distribución de x")
fig_box = px.box(df, x="cat", y="y", points="all", title="Distribución y por categoría")
fig_bar, fig_hist, fig_box
```

## 4) Guardar a HTML

```

ruta_html = "/mnt/data/plotly_demo_visualizacion.html"
fig_bar.write_html(ruta_html, include_plotlyjs="cdn", full_html=True)
ruta_html
```

## ✅ Autotests 13

```

from pathlib import Path
import plotly.graph_objs as go

def check(cond, ok, fail):
    if cond: print("✔", ok)
    else: raise AssertionError("✘ " + fail)

# Test 1: dataset correcto
check(df.shape[0] == 400 and set(df.columns) == {"x","y","cat","tam"},
      "Dataset df correcto (400 filas, columnas x,y,cat,tam).",
      "Estructura de df inesperada.")

# Test 2: figuras creadas
check(isinstance(fig_scatter, go.Figure) and isinstance(fig_bar, go.Figure), 
      "Figuras Plotly creadas (scatter y bar).",
      "Faltan figuras base.")

# Test 3: HTML exportado
check(Path(ruta_html).exists(), f"HTML exportado en {ruta_html}.", "No se encontró el HTML exportado.")
```
