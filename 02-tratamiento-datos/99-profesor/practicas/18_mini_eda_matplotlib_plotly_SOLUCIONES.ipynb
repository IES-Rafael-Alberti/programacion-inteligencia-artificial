# 14 - Mini‑EDA y Storytelling — **Soluciones + Autotests**


Incluye: mini‑EDA con dos gráficos, matriz de correlación y **resumen narrativo** exportable.


## 1) Dataset y preguntas guía

```

import numpy as np, pandas as pd
rng = np.random.default_rng(7)
n = 500
df = pd.DataFrame({
    "edad": rng.integers(18, 75, size=n),
    "ingresos": rng.normal(26000, 9000, size=n),
    "ciudad": rng.choice(["Cádiz","Sevilla","Málaga","Córdoba"], size=n),
    "compra": rng.choice([0,1], size=n, p=[0.45, 0.55])
})
df.shape
```

## 2) Mini‑EDA: resúmenes y gráficos

```

import matplotlib.pyplot as plt

resumen = df.describe(numeric_only=True)
corr = df[["edad","ingresos","compra"]].corr(numeric_only=True)

fig1, ax = plt.subplots()
ax.scatter(df["edad"], df["ingresos"], alpha=0.5)
ax.set_title("Edad vs Ingresos")
ax.set_xlabel("Edad"); ax.set_ylabel("Ingresos")
plt.show()

fig2, ax = plt.subplots()
df.groupby("ciudad")["ingresos"].mean().sort_values().plot(kind="bar", ax=ax)
ax.set_title("Ingresos medios por ciudad")
plt.show()
```

## 3) Storytelling y exportación

```

from textwrap import dedent

narrativa = dedent(f'''
### Hallazgos clave
- Ingresos medios: {df["ingresos"].mean():.1f} (std {df["ingresos"].std():.1f}).
- Correlación edad‑ingresos: {corr.loc["edad","ingresos"]:.2f}.
- Ciudad con mayor ingreso medio: {df.groupby("ciudad")["ingresos"].mean().idxmax()}.
- Tasa de compra global: {df["compra"].mean():.2%}.
''')

ruta_md = "/mnt/data/mini_eda_storytelling_visualizacion.md"
with open(ruta_md, "w", encoding="utf-8") as f:
    f.write(narrativa)
ruta_md
```

## ✅ Autotests 14

```

from pathlib import Path

def check(cond, ok, fail):
    if cond: print("✔", ok)
    else: raise AssertionError("✘ " + fail)

# Test 1: dimensiones dataset
check(df.shape[0] == 500 and set(df.columns) == {"edad","ingresos","ciudad","compra"},
      "Dataset df correcto (500 filas, columnas esperadas).",
      "Estructura de df inesperada.")

# Test 2: resúmenes y correlaciones creados
check('resumen' in globals() and 'corr' in globals(), "resumen y corr creados.", "Faltan resúmenes/correlación.")

# Test 3: narrativa exportada
check(Path(ruta_md).exists() and len(open(ruta_md, 'r', encoding='utf-8').read()) > 0,
      f"Narrativa exportada en {ruta_md}.", "No se produjo narrativa exportada.")
```
