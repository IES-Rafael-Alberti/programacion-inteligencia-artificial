```python
import os
from pathlib import Path
from dotenv import load_dotenv

def find_repository_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "pixi.toml").is_file() and (candidate / "03-machine-learning").is_dir():
            return candidate
    raise FileNotFoundError("No se encontró la raíz del repositorio PIA desde el directorio actual.")

repository_root = find_repository_root(Path.cwd().resolve())
project_root = repository_root / "03-machine-learning/03-practicas/laboratorios/Lab1/mi_dataset_peliculas"
env_path = project_root / ".env"
if not env_path.is_file():
    raise FileNotFoundError(f"No se encontró el archivo local esperado: {env_path}")
load_dotenv(dotenv_path=env_path)
TMDB_API_KEY = os.getenv("TMDB_API_KEY")
OMDB_API_KEY = os.getenv("OMDB_API_KEY")
if not TMDB_API_KEY or not OMDB_API_KEY:
    raise RuntimeError("Falta configurar TMDB_API_KEY u OMDB_API_KEY en .env.")
print("Claves API cargadas correctamente.")
```
