# Entidades con spaCy - Soluciones

```
import spacy
nlp = spacy.load('es_core_news_sm')
doc = nlp("Apple planea abrir una nueva oficina en Barcelona en 2024.")
[(ent.text, ent.label_) for ent in doc.ents]
```

## Tests automáticos


```
import spacy
nlp = spacy.load('es_core_news_sm')
doc = nlp("Apple planea abrir una nueva oficina en Barcelona en 2024.")
ents = [ent.text for ent in doc.ents]
assert 'Apple' in ents
assert 'Barcelona' in ents
```
