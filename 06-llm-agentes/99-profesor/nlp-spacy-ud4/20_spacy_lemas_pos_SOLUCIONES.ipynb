# Lematización y análisis morfosintáctico - Soluciones

```
import spacy
nlp = spacy.load('es_core_news_sm')
text = "Los niños estaban corriendo rápidamente por el parque."
doc = nlp(text)
[(token.text, token.lemma_, token.pos_) for token in doc]
```

## Tests automáticos


```
import spacy
nlp = spacy.load('es_core_news_sm')
doc = nlp("Los niños estaban corriendo rápidamente por el parque.")
assert 'correr' in [token.lemma_ for token in doc]
assert any(token.pos_ == 'VERB' for token in doc)
```
