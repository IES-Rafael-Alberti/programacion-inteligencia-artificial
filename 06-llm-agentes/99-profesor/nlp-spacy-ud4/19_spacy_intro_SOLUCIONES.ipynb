# Introducción a spaCy - Soluciones
Este notebook contiene las soluciones de los ejercicios propuestos.

```
import spacy
nlp = spacy.load('es_core_news_sm')
text = "Hola, me llamo Ana y vivo en Madrid."
doc = nlp(text)
[(token.text, token.pos_, token.dep_) for token in doc]
```

## Tests automáticos


```
import spacy
nlp = spacy.load('es_core_news_sm')
text = "Hola, me llamo Ana y vivo en Madrid."
doc = nlp(text)

assert any(token.text == 'Ana' for token in doc)
assert any(token.pos_ == 'PROPN' for token in doc)
```
