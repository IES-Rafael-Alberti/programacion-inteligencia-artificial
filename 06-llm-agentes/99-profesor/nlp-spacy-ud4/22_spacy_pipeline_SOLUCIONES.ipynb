# Mini Pipeline NLP con spaCy - Soluciones

```
import spacy
import matplotlib.pyplot as plt
from collections import Counter

nlp = spacy.load('es_core_news_sm')

text = "El aprendizaje automático y la inteligencia artificial están revolucionando el mundo."
doc = nlp(text)

lemmas = [t.lemma_ for t in doc if not t.is_stop and t.is_alpha]
frecuencias = Counter(lemmas)
frecuencias
```

```
plt.bar(frecuencias.keys(), frecuencias.values())
plt.show()
```
