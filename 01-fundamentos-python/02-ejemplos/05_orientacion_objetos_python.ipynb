## 1. Clases e instancias

Una clase describe una estructura y un comportamiento; cada instancia es un objeto concreto.

```python
class Alumno:
    pass

ana = Alumno()
bruno = Alumno()
print(type(ana).__name__, ana is bruno)
```

## 2. `__init__`, atributos y métodos

El inicializador recibe los datos necesarios. Los métodos agrupan operaciones que tienen sentido para el objeto.

```python
class Alumno:
    def __init__(self, nombre, curso):
        self.nombre = nombre
        self.curso = curso
        self.calificaciones = []

    def matricular(self, curso):
        self.curso = curso

    def añadir_calificacion(self, nota):
        self.calificaciones.append(nota)

    def media(self):
        return sum(self.calificaciones) / len(self.calificaciones) if self.calificaciones else 0.0

ana = Alumno('Ana', '1º IA')
ana.añadir_calificacion(8)
ana.añadir_calificacion(9)
print(ana.nombre, ana.media())
```

## 3. Composición

La composición expresa una relación de tipo «tiene»: un `Curso` contiene alumnado. Cada clase conserva una responsabilidad sencilla.

```python
class Curso:
    def __init__(self, nombre):
        self.nombre = nombre
        self.alumnos = []

    def matricular(self, alumno):
        self.alumnos.append(alumno)
        alumno.matricular(self.nombre)

curso = Curso('1º IA')
curso.matricular(ana)
curso.matricular(Alumno('Bruno', ''))
print(curso.nombre, [a.nombre for a in curso.alumnos])
```

## 4. Herencia y `super()`

Una subclase especializa una clase base. `super()` permite reutilizar su inicialización y comportamiento sin copiarlo.

```python
class AlumnoBecado(Alumno):
    def __init__(self, nombre, curso, porcentaje):
        super().__init__(nombre, curso)
        self.porcentaje_beca = porcentaje

    def resumen_beca(self):
        return f'{self.nombre}: beca del {self.porcentaje_beca}%'

becado = AlumnoBecado('Clara', '1º IA', 50)
print(becado.resumen_beca(), becado.curso)
```

## 5. Métodos especiales

Python usa protocolos: `__repr__` define una representación útil, `__len__` permite `len(objeto)` y `__getitem__` permite indexar.

```python
class CatalogoCursos:
    def __init__(self, cursos):
        self.cursos = list(cursos)

    def __repr__(self):
        return f'CatalogoCursos({self.cursos!r})'

    def __len__(self):
        return len(self.cursos)

    def __getitem__(self, indice):
        return self.cursos[indice]

catalogo = CatalogoCursos(['Python', 'NumPy', 'POO'])
print(catalogo)
print(len(catalogo), catalogo[1], catalogo[-1])
```

## 6. `@dataclass`

Para objetos centrados en datos, `dataclass` genera automáticamente métodos habituales como `__init__` y `__repr__`.

```python
from dataclasses import dataclass

@dataclass
class RegistroCurso:
    alumno: str
    curso: str
    nota: float = 0.0

registro = RegistroCurso('Ana', '1º IA', 8.5)
print(registro)
```
