## Productos de matriz y vector: `@` / `matmul`

```python
import numpy as np

A = np.array([[1., 2., 3.],
              [4., 5., 6.]])          # (2, 3)
v = np.array([1., 0., -1.])           # (3,)
b = A @ v                             # (2,)
print(b.shape, b)
```

```python
B = np.array([[1., 0.],
              [0., 1.],
              [2., 2.]])              # (3, 2)
C = A @ B                             # (2, 2)
print(C)

# Ojo: * es elemento a elemento, NO producto matricial
print('producto elemento a elemento (no matricial):')
print(A[:, :2] * A[:, :2])
```

## Normas y distancias: `np.linalg.norm`

```python
p = np.array([3., 4.])
print('norma L2:', np.linalg.norm(p))
print('norma L1:', np.linalg.norm(p, ord=1))
print('norma inf:', np.linalg.norm(p, ord=np.inf) )
```

```python
X = np.array([[0., 0.],
              [3., 4.],
              [6., 8.]])
print('normas por fila:', np.linalg.norm(X, axis=1))

a = np.array([1., 2.])
bp = np.array([4., 6.])
print('distancia a-b:', np.linalg.norm(a - bp))
```

## Resolver sistemas: `solve` (mejor que `inv`)

```python
A = np.array([[2., 1.],
              [1., 3.]])
bl = np.array([8., 13.])
x = np.linalg.solve(A, bl)
print('solución:', x)
print('A @ x aprox. b:', A @ x)
```

## Inversa, determinante y traza

```python
print('inv:')
print(np.linalg.inv(A))
print('det:', np.linalg.det(A))
print('traza:', np.trace(A))

# Comprobar A @ inv(A) ≈ identidad
print('A @ inv(A):')
print(A @ np.linalg.inv(A))
```

## Autovalores y autovectores: `eig` / `eigh`

```python
A = np.array([[4., -2.],
              [1.,  1.]])
w, v = np.linalg.eig(A)
print('autovalores:', w)
print('autovectores (columnas):\n', v)
```

## SVD y conexión con PCA

```python
rng = np.random.default_rng(0)
X = rng.normal(size=(50, 4))
U, S, Vt = np.linalg.svd(X)
print('U:', U.shape, 'S:', S.shape, 'Vt:', Vt.shape)
print('valores singulares:', S)

k = 2
X_aprox = (U[:, :k] * S[:k]) @ Vt[:k, :]
print('error de reconstrucción k=2:', np.linalg.norm(X - X_aprox))
```

## Normalización de filas con `keepdims=True`

```python
X = np.random.default_rng(1).normal(size=(5, 3))
norms = np.linalg.norm(X, axis=1, keepdims=True)
X_norm = X / norms
print('normas resultantes:', np.linalg.norm(X_norm, axis=1))
```

## Condición de una matriz

```python
A = np.array([[1., 1.],
              [1., 1. + 1e-8]])
print('número de condición:', np.linalg.cond(A))
```
