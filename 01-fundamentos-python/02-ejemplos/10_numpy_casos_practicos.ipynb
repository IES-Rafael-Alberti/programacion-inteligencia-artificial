# Taller de casos prácticos con NumPy

Ejemplos cortos para reforzar ejes, máscaras, broadcasting, álgebra lineal y análisis. Ejecuta las celdas en orden.

```python
import numpy as np
rng = np.random.default_rng(2026)
```

## 1. Ejes, reducciones y `keepdims`

```python
M = np.arange(12, dtype=float).reshape(3, 4)
print("M.shape:", M.shape)
print("Total:", M.sum())
print("Por columna:", M.sum(axis=0))
print("Por fila:", M.sum(axis=1))
media_col = M.mean(axis=0, keepdims=True)
print("Forma de la media conservando eje:", media_col.shape)
print("Centrada por columna:\n", M - media_col)
```

## 2. Indexado avanzado y máscaras booleanas

```python
x = np.array([10, 20, 30, 40, 50])
print("Índices seleccionados:", x[[0, 2, 4]])
print("Mayores que 25:", x[x > 25])
print("Entre 15 y 45:", x[(x > 15) & (x < 45)])
x[x < 30] = 0
print("Tras asignar con máscara:", x)
```

## 3. `where`, `select`, `clip` y ordenación

```python
notas = np.array([3.5, 5.0, 6.8, 8.9, 9.5])
print(np.where(notas >= 5, "apta", "revisar"))
rangos = np.select([notas < 5, notas < 7, notas < 9], ["insuficiente", "aprobado", "notable"], default="sobresaliente")
print(rangos)
print("Limitadas a [4,8]:", np.clip(notas, 4, 8))
print("Índice de la mayor:", notas.argmax())
print("Orden de índices:", notas.argsort())
```

## 4. Valores ausentes

Compara una reducción normal con la función `nan*` e imputa solo las posiciones ausentes.

```python
mediciones = np.array([12.0, np.nan, 15.0, 9.0, np.nan])
print("Media normal:", np.mean(mediciones))
media_valida = np.nanmean(mediciones)
mediciones[np.isnan(mediciones)] = media_valida
print("Media válida:", media_valida, "| array imputado:", mediciones)
```

## 5. Álgebra lineal: `*` elemento a elemento y `@` matricial

```python
A = np.array([[2., 1.], [1., 3.]])
B = np.array([[1., 2.], [3., 4.]])
print("Elemento a elemento:\n", A * B)
print("Producto matricial:\n", A @ B)
b = np.array([5., 7.])
sol = np.linalg.solve(A, b)
print("Solución Ax=b:", sol, "| comprobación:", A @ sol)
print("Determinante:", np.linalg.det(A))
```

## 6. Concatenar, apilar y crear rejillas

```python
u = np.array([1, 2, 3]); v = np.array([10, 20, 30])
print("concatenate:", np.concatenate([u, v]))
print("stack axis=0:\n", np.stack([u, v]))
print("stack axis=1:\n", np.stack([u, v], axis=1))
print("producto exterior:\n", u[:, None] * v[None, :])
```

## 7. Broadcasting aplicado: escalar, filas y columnas

```python
X = np.array([[10., 100.], [20., 200.], [30., 300.]])
offset = np.array([1., 2.])
print("Suma por columna:\n", X + offset)
pesos_fila = np.array([1., 0.5, 2.])
print("Escalado por fila:\n", X * pesos_fila[:, None])
print("Formas:", X.shape, offset.shape, pesos_fila[:, None].shape)
```

### Normalizar cada columna al intervalo [0, 1]

```python
X = np.array([[2., 10., 100.], [4., 20., 200.], [6., 30., 300.]])
xmin = X.min(axis=0, keepdims=True)
xmax = X.max(axis=0, keepdims=True)
X01 = (X - xmin) / (xmax - xmin)
print(X01)
print("Mínimos:", X01.min(axis=0), "máximos:", X01.max(axis=0))
```

## 8. Matriz de distancias por parejas con broadcasting

```python
puntos = np.array([[0., 0.], [3., 4.], [6., 8.]])
diferencias = puntos[:, None, :] - puntos[None, :, :]
D = np.sqrt((diferencias ** 2).sum(axis=-1))
print(D)
print("Simétrica:", np.allclose(D, D.T), "| diagonal:", np.diag(D))
```

## 9. Aleatoriedad reproducible y correlación

```python
rng = np.random.default_rng(42)
altura = rng.normal(170, 10, size=200)
peso = 0.4 * altura + rng.normal(0, 8, size=200)
X = np.column_stack([altura, peso])
print("Medias:", X.mean(axis=0).round(2))
print("Correlación:\n", np.corrcoef(X, rowvar=False).round(2))
rng2 = np.random.default_rng(42)
print("Reproducible:", np.array_equal(rng.normal(size=5), rng2.normal(size=5)))
```

## 10. Vectorización y medición sencilla

Los tiempos dependen del equipo; compara la misma operación y los mismos datos.

```python
import time
x = np.arange(100_000, dtype=np.float64)
t0 = time.perf_counter()
y_loop = [valor * valor + 2 * valor + 1 for valor in x]
t_python = time.perf_counter() - t0
t0 = time.perf_counter()
y_numpy = x * x + 2 * x + 1
t_numpy = time.perf_counter() - t0
print(f"Python: {t_python:.4f}s | NumPy: {t_numpy:.4f}s")
print("Equivalentes:", np.allclose(y_loop, y_numpy))
```
