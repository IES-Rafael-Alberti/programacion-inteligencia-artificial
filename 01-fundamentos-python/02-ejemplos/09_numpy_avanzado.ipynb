```python
import numpy as np
import time

rng = np.random.default_rng(2026)
print(np.__version__)
```

## Tipo, forma y ejes
Observa cómo `dtype`, `shape` y `axis` describen el array.

```python
a = np.arange(12, dtype=np.int32).reshape(3, 4)
print(a)
print("shape:", a.shape, "dtype:", a.dtype, "bytes:", a.nbytes)
print("por columnas:", a.sum(axis=0))
print("por filas:", a.sum(axis=1))
```

## Máscaras, `where` y `select`

```python
notas = np.array([4.5, 6.0, 8.5, 9.7])
print(notas[(notas >= 5) & (notas < 9)])
print(np.where(notas >= 5, "apto", "no apto"))
print(np.select([notas < 5, notas < 7, notas < 9], ["insuficiente", "aprobado", "notable"], default="sobresaliente"))
```

## Vistas y copias

```python
base = np.arange(6)
view = base[1:4]
view[0] = 99
copy = base[1:4].copy()
copy[0] = -1
print(base, copy)
print(np.shares_memory(base, view), np.shares_memory(base, copy))
```

## Aleatoriedad reproducible y vectorización

```python
sample = rng.normal(size=1000).astype(np.float32)
print(sample[:5], sample.dtype)
x = np.arange(100_000, dtype=np.float64)
t0 = time.perf_counter()
y = x * x + 2 * x + 1
print(y[:3], time.perf_counter() - t0)
```
