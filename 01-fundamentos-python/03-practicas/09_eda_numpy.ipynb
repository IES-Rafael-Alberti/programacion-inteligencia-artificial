## 🧾 Contexto del dataset

Se trabajará con un dataset sintético que simula **información de empleados**:

* ID
* Edad
* Salario mensual
* Años de experiencia
* Horas de trabajo por semana
* Nivel educativo (0 a 4)
* Satisfacción (0–10, con valores faltantes o atípicos)

Generaremos los datos con `np.random`, así todos tendrán el mismo formato.

```python
import numpy as np

np.random.seed(42)
n = 1000

ids = np.arange(1, n+1)
edad = np.random.randint(18, 65, size=n)
salario = np.random.normal(2500, 800, size=n)
experiencia = np.clip(edad - np.random.randint(18, 30, size=n), 0, None)
horas = np.random.normal(40, 5, size=n)
nivel = np.random.randint(0, 5, size=n)
satisfaccion = np.random.normal(7, 2, size=n)

# Añadimos valores atípicos y NaN
salario[np.random.randint(0, n, 10)] *= 4
satisfaccion[np.random.randint(0, n, 15)] = np.nan

data = np.column_stack([ids, edad, salario, experiencia, horas, nivel, satisfaccion])
print(data) 
```

### ⚠️ **Tarea 5: Detección de outliers**

1. Calcula los **cuartiles (Q1 y Q3)** del salario.
2. Calcula el **rango intercuartílico (IQR)**.
3. Encuentra los valores que están fuera de `Q1 - 1.5*IQR` o `Q3 + 1.5*IQR`.
4. Muestra cuántos outliers hay.

➡️ Objetivo: detección numérica de valores atípicos con operaciones vectorizadas.

---



```python
salarios = data[:, 2]
# Calcular Q1 y Q3
q1, q3 = np.nanpercentile(salarios, [25, 75])

q1,q3
```

```python
# calcular rango intercuartílico
iqr = q3 -q1
iqr
```

```python
# 3. Límites y detección de outliers
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

limite_inferior, limite_superior
```

```python
# calcular outliers
outliers = (salarios < limite_inferior) | (salarios > limite_superior)
valores_atipicos = salarios[outliers]
```

```python
# 4. Número de outliers
numero_outliers = np.sum(outliers)
len(valores_atipicos), numero_outliers
```

```python
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
print("Valores atípicos:", valores_atipicos)
print("Número de outliers:", numero_outliers)
```

### 🧰 **Tarea 6: Sustitución o corrección de outliers**

1. Sustituye los salarios outliers por el valor del **percentil 95**.
2. Verifica que ya no hay salarios fuera del rango esperado.

➡️ Objetivo: aplicar `np.where` y percentiles para limpiar valores extremos.

---



```python
# Percentil 95
p95 = np.nanpercentile(salarios, 95)
```

```python
# Sustituir outliers usando np.where
salarios_limpios = np.where(outliers, p95, salarios)
```

```python
# Actualizar la columna de salarios
data[:, 2] = salarios_limpios
```

```python
# Verificar
outliers_restantes = (
    (data[:, 2] < limite_inferior) |
    (data[:, 2] > limite_superior)
)
```

```python
print("Percentil 95:", p95)
print("Outliers restantes:", np.sum(outliers_restantes))
```

```python
p95_ajustado = min(p95, limite_superior)
data[:, 2] = np.where(outliers, p95_ajustado, salarios)
```

### 🧩 **Tarea 8: Normalización y escalado**

1. Normaliza la variable **salario** entre 0 y 1.
2. Estandariza (z-score) la variable **horas**.
3. Comprueba media ~0 y std ~1.

➡️ Objetivo: practicar transformaciones de escala (útil para ML más adelante).

---



```python
#preparación datos
# Columnas:
# Salario = 2
# Horas semanales = 4

salarios = data[:, 2]
horas = data[:, 4]
```

```python
# 1. Normalización Min-Max entre 0 y 1
salario_min = np.min(salarios)
salario_max = np.max(salarios)

salario_normalizado = (
    (salarios - salario_min) /
    (salario_max - salario_min)
)
```

```python
#comprobar que min-max normalization todos los datos están entre 0 y 1
esta_normalizado = np.all(
    (salario_normalizado >= 0) &
    (salario_normalizado <= 1)
)

print(esta_normalizado)

# otra comprobación usando extremos
print("Mínimo:", np.min(salario_normalizado))
print("Máximo:", np.max(salario_normalizado))
```

```python
# 2. Estandarización Z-score de las horas
media_horas = np.mean(horas)
std_horas = np.std(horas)

horas_estandarizadas = (horas - media_horas) / std_horas
```

```python
# 3. Comprobación
print("Media:", np.mean(horas_estandarizadas))
print("Desviación estándar:", np.std(horas_estandarizadas))
```

### 🧮 **Tarea 9: Correlaciones simples**

1. Calcula la **correlación** entre salario y experiencia.
2. Calcula la correlación entre edad y satisfacción.

➡️ Objetivo: usar `np.corrcoef` e interpretar relaciones lineales.

---



```python
# obtener datos
# Columnas:
# Edad = 1
# Salario = 2
# Experiencia = 3
# Satisfacción = 6

salarios = data[:, 2]
experiencia = data[:, 3]
edades = data[:, 1]
satisfaccion = data[:, 6]
```

```python
# Correlación salario-experiencia
corr_salario_experiencia = np.corrcoef(
    salarios,
    experiencia
)[0, 1]
```

```python
# Correlación edad-satisfacción
corr_edad_satisfaccion = np.corrcoef(
    edades,
    satisfaccion
)[0, 1]
```

```python
print("Salario y experiencia:",
      corr_salario_experiencia)

print("Edad y satisfacción:",
      corr_edad_satisfaccion)
```

Interpretación:
- Cerca de 1: relación lineal positiva fuerte.
- Cerca de -1: relación lineal negativa fuerte.
- Cerca de 0: poca o ninguna relación lineal.
np.corrcoef devuelve una matriz de correlación; [0, 1] selecciona la correlación entre las dos variables.

### 🧾 **Tarea 10: Estadísticas generales**

1. Muestra para cada columna:

   * Media
   * Desviación estándar
   * Valor mínimo
   * Valor máximo
   * Porcentaje de valores faltantes

➡️ Objetivo: preparar el pensamiento para `df.describe()` y `df.info()` de Pandas.

```python
nombres = [
    "ID",
    "Edad",
    "Salario",
    "Experiencia",
    "Horas",
    "Nivel educativo",
    "Satisfacción"
]

for i, nombre in enumerate(nombres):
    columna = data[:, i]

    faltantes = np.isnan(columna).sum()
    porcentaje_faltantes = faltantes / len(columna) * 100

    print(f"\n{nombre}")
    print(f"  Media:              {np.nanmean(columna):.2f}")
    print(f"  Desviación estándar: {np.nanstd(columna):.2f}")
    print(f"  Mínimo:             {np.nanmin(columna):.2f}")
    print(f"  Máximo:             {np.nanmax(columna):.2f}")
    print(f"  Valores faltantes:  {porcentaje_faltantes:.2f}%")
```
