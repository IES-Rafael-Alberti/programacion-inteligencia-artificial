## 2) Mini‑ejercicio: crea dos variables y súmalas

**Enunciado**: Crea `a=7` y `b=5`, guarda la suma en `suma_ab`.

### ✅ Solución

```
a = 7
b = 5
suma_ab = a + b
suma_ab
```

---

## ✅ Autotests 01

```
def check(cond, ok, fail):
    if cond: print("✔", ok)
    else: raise AssertionError("✘ " + fail)

check('suma_ab' in globals(), "Variable 'suma_ab' creada.", "Falta la variable 'suma_ab'.")
check(suma_ab == 12, "La suma es correcta (=12).", "La suma no es 12.")
```
