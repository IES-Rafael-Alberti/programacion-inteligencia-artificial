# 01 - Introducción a Python — **Soluciones**

Primeros pasos con Python: variables, tipos, operaciones básicas.

```

# Variables y tipos
nombre = "Juan"
edad = 25
pi = 3.1416
es_estudiante = True

print(type(nombre), type(edad), type(pi), type(es_estudiante))

# Operaciones
suma = 7 + 3
resta = 10 - 4
producto = 5 * 6
division = 8 / 3
division_entera = 8 // 3
modulo = 8 % 3
potencia = 2 ** 5

suma, resta, producto, division, division_entera, modulo, potencia
```
