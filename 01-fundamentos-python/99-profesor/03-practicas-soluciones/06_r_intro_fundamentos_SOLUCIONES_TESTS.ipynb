# 06 - Introducción y Fundamentos de R — **Soluciones + Tests Autocorregibles**

Ejecuta las celdas de abajo para comprobar automáticamente tus respuestas.

```

# Variables de prueba
x <- 10
y <- 5
suma <- x + y

stopifnot(suma == 15)
print("✅ Test 1 superado: suma correcta")
```

```

# Vectores
v <- c(1,2,3,4,5)
stopifnot(length(v) == 5)
print("✅ Test 2 superado: longitud correcta del vector")
```

```

# Data frames
df <- data.frame(nombre=c("Ana","Luis","Marta"),
                 edad=c(21,25,30))

stopifnot(ncol(df) == 2)
stopifnot("nombre" %in% names(df))
print("✅ Test 3 superado: data frame correcto")
```
