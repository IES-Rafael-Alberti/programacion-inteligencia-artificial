## Ejercicio 1 — Array 4x4 y diagonal

```
import numpy as np

A = np.arange(16).reshape(4,4)
diag = np.diag(A)           # o A[np.arange(4), np.arange(4)]
A, diag
```

## Ejercicio 2 — 1000 normales y estadísticos

```
rng = np.random.default_rng(42)
muestra = rng.normal(0, 1, 1000)
media = muestra.mean()
desv = muestra.std(ddof=0)  # poblacional
media, desv
```

## Ejercicio 3 — Cuadrados sin bucles

```
x = np.array([1,2,3,4,5])
cuadrados = x**2
cuadrados
```

## ✅ Autotests 06

```
def check(cond, msg_ok, msg_fail):
    if cond:
        print("✔", msg_ok)
    else:
        raise AssertionError("✘ " + msg_fail)

# Test 1: forma de A y diagonal correcta
check(A.shape == (4,4), "A tiene forma 4x4.", "A no es 4x4.")
check((diag == np.array([0,5,10,15])).all(), "Diagonal correcta extraída.", "Diagonal incorrecta.")

# Test 2: tamaño muestra y estadísticos razonables
check(len(muestra) == 1000, "Muestra de tamaño 1000.", "Tamaño de muestra incorrecto.")
check(abs(media) < 0.2 and 0.8 < desv < 1.2, "Media ~0 y sigma ~1 (aprox).", "Estadísticos fuera de rango razonable.")

# Test 3: cuadrados correctos
check((cuadrados == np.array([1,4,9,16,25])).all(), "Vector de cuadrados correcto.", "Cuadrados incorrectos.")
```
