## Ejercicio 1
Array 2x3 + vector de longitud 3 con broadcasting y explicación de la forma resultante.

```
import numpy as np

X = np.array([[1,2,3],[4,5,6]])   # forma (2,3)
v = np.array([10,20,30])          # forma (3,)
Y = X + v                         
Y, X.shape, v.shape, Y.shape  # v se 'expande' a (1,3) y luego a (2,3)
```

**Explicación:** El vector `(3,)` se alinea con el eje de columnas y se difunde a `(2,3)` para sumar elemento a elemento.

## Ejercicio 2
Comparar el tiempo de `a*b + a` con bucles vs vectorizado.

```
import time
n = 400_000
a = np.random.rand(n)
b = np.random.rand(n)

t0 = time.time()
out_loop = np.empty_like(a)
for i in range(n):
    out_loop[i] = a[i]*b[i] + a[i]
t1 = time.time()

t2 = time.time()
out_vec = a*b + a
t3 = time.time()

(t1-t0, t3-t2, np.allclose(out_loop, out_vec))
```

## Ejercicio 3
Resuelve un sistema lineal 3x3 y verifica el resultado.

```
M = np.array([[2., -1., 0.],
              [ -1., 2., -1.],
              [0., -1., 2.]])
b = np.array([1., 0., 1.])

x = np.linalg.solve(M, b)
residuo = M @ x - b
x, residuo, np.allclose(M @ x, b)
```
