Objetivos:

## Ejercicio A — Condicional y bucles

### ✅ Solución

```
edad = 17
if edad >= 18:
    mensaje = "Mayor de edad"
elif edad >= 16:
    mensaje = "Puede votar con autorización"
else:
    mensaje = "Menor de edad"

nums_for = [i for i in range(1,6)]
contador = 0
nums_while = []
while contador < 5:
    nums_while.append(contador)
    contador += 1

mensaje, nums_for, nums_while
```

## Ejercicio B — Funciones

### ✅ Solución

```
def saludar(nombre):
    return f"Hola, {nombre}"

def sumar(a, b):
    return a + b

saludo_demo = saludar("Ana")
suma_demo = sumar(4, 7)
saludo_demo, suma_demo
```

## ✅ Autotests 03

```
def check(cond, ok, fail):
    if cond: print("✔", ok)
    else: raise AssertionError("✘ " + fail)

check(mensaje in {"Mayor de edad","Puede votar con autorización","Menor de edad"},
      "Mensaje definido correctamente.", "Mensaje no válido.")
check(nums_for == [1,2,3,4,5] and nums_while == [0,1,2,3,4],
      "Listas de bucles correctas.", "Listas de bucles incorrectas.")
check(callable(saludar) and saludar("Luis") == "Hola, Luis", "Función saludar correcta.", "saludar no funciona.")
check(callable(sumar) and sumar(2,3) == 5, "Función sumar correcta.", "sumar no funciona.")
```
