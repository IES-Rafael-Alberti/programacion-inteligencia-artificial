# 05 - Repaso y ejercicios básicos — **Soluciones + Autotests**


Ejercicios integradores sobre variables, control de flujo, funciones y estructuras.


## Ejercicio 1 — Variables y operadores

### ✅ Solución

```
a, b = 7, 3
suma = a + b
resta = a - b
producto = a * b
suma, resta, producto
```

## Ejercicio 2 — Condicionales

### ✅ Solución

```
edad = 17
if edad >= 18:
    estado = "adulto"
else:
    estado = "menor"
estado
```

## Ejercicio 3 — Bucle for

### ✅ Solución

```
lista_1_10 = [i for i in range(1,11)]
lista_1_10[:5], lista_1_10[-3:]
```

## Ejercicio 4 — Función con retorno

### ✅ Solución

```
def cuadrado(x):
    return x**2
c4 = cuadrado(4)
c4
```

## Ejercicio 5 — Diccionarios y listas

### ✅ Solución

```
alumnos = [
    {"nombre":"Ana","nota":7},
    {"nombre":"Luis","nota":4},
    {"nombre":"Marta","nota":9},
]
aprobados = [(al["nombre"], al["nota"]>=5) for al in alumnos]
aprobados[:3]
```

---

## ✅ Autotests 05

```
def check(cond, ok, fail):
    if cond: print("✔", ok)
    else: raise AssertionError("✘ " + fail)

check((suma,resta,producto)==(10,4,21), "Operadores correctos.", "Resultados aritméticos incorrectos.")
check(estado in {"adulto","menor"}, "Condicional correcto.", "Variable 'estado' no válida.")
check(lista_1_10[0]==1 and lista_1_10[-1]==10 and len(lista_1_10)==10, "Bucle for correcto.", "Lista 1..10 incorrecta.")
check(callable(cuadrado) and c4==16, "Función cuadrado correcta.", "Función cuadrado falla.")
check(len(aprobados)==3 and aprobados[0][1] is True and aprobados[1][1] is False, "Procesamiento de alumnos correcto.", "Error en lista de aprobados.")
```
