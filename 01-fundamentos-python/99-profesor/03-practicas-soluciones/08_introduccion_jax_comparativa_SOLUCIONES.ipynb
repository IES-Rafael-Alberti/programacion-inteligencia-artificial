## Ejercicio 1
Implementa `f(x) = x^3 + 2x` y calcula `f'(3)` con `jax.grad`.

```
try:
    import jax, jax.numpy as jnp
    def f(x):
        return x**3 + 2*x
    df = jax.grad(f)
    df(3.0)
except Exception as e:
    print("JAX no disponible o error:", e)
```

## Ejercicio 2
Usa `jax.jit` para acelerar `a*b + a**2` (compara 1ª vs 2ª ejecución).

```
try:
    import time, numpy as np
    import jax, jax.numpy as jnp

    @jax.jit
    def g(a, b):
        return a*b + a**2

    a_np = np.random.rand(200_000)
    b_np = np.random.rand(200_000)
    a_j = jnp.array(a_np); b_j = jnp.array(b_np)

    t0 = time.time()
    out1 = g(a_j, b_j).block_until_ready()
    t1 = time.time()
    out2 = g(a_j, b_j).block_until_ready()
    t2 = time.time()

    (t1-t0, t2-t1)
except Exception as e:
    print("JAX no disponible o error:", e)
```

## Ejercicio 3
Con `vmap`, aplica `x^2 + 1` a un vector de 10 elementos.

```
try:
    import jax, jax.numpy as jnp
    def h(x):
        return x**2 + 1
    hv = jax.vmap(h)
    xs = jnp.arange(10.0)
    hv(xs)
except Exception as e:
    print("JAX no disponible o error:", e)
```
