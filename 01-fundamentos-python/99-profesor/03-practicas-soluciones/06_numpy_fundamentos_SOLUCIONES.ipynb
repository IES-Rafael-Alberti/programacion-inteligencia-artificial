
##### 4) Estandarizar columnas (z-score por columna)
Enunciado:
Crea una matriz aleatoria A de tamaño 5×4 con np.random.randn. Calcula, para cada columna, su media y desviación estándar y estandariza: Z=(A−μ)/σ. Verifica que la media de cada columna de Z es ~0 y su std ~1.



```python
# Solución:

import numpy as np
np.random.seed(0)
A = np.random.randn(5, 4)
mu = A.mean(axis=0)
sd = A.std(axis=0, ddof=0)
Z = (A - mu) / sd
print(Z.mean(axis=0), Z.std(axis=0))
```

#### 5) Borde a 1, interior a 0
Enunciado:
Crea una matriz M 6×6 con los números del 0 al 35 (con arange+reshape). Pon el borde de M a 1 y el interior a 0.



```python
# 5) Solución:

M = np.arange(36).reshape(6,6).astype(float)
M[:] = 0
M[0,:] = M[-1,:] = M[:,0] = M[:,-1] = 1
print(M)
# o alternativamente:
M = np.ones((6,6))
M[1:-1, 1:-1] = 0
print(M)

# otra opción:
M.fill(0)
M[[0,-1], :] = 1
M[:, [0,-1]] = 1
```

```python
# 5) Solución:

M = np.arange(36).reshape(6,6).astype(float)
M[:] = 0
M[0,:] = M[-1,:] = M[:,0] = M[:,-1] = 1
print(M)

# Sin slicing - opción 1: usando np.pad
M = np.pad(np.zeros((4,4)), pad_width=1, constant_values=1)
print("\nCon np.pad:")
print(M)

# Sin slicing - opción 2: usando máscaras booleanas
M = np.arange(36).reshape(6,6)
i, j = np.indices((6,6))
mask_borde = (i == 0) | (i == 5) | (j == 0) | (j == 5)
M = np.where(mask_borde, 1, 0)
print("\nCon máscaras booleanas:")
print(M)

# Sin slicing - opción 3: usando np.minimum para calcular distancia al borde
M = np.arange(36).reshape(6,6)
i, j = np.indices((6,6))
dist_borde = np.minimum(np.minimum(i, 5-i), np.minimum(j, 5-j))
M = (dist_borde == 0).astype(int)
print("\nCon distancia al borde:")
print(M)
```

11) Concatenar con padding
Enunciado:
Dados a = [1,2,3,4], b = [5,6], c = [7,8,9], construye una matriz 3×4 donde cada fila sea a, b y c padded con ceros a la derecha.

Solución:

a = np.array([1,2,3,4])
b = np.array([5,6])
c = np.array([7,8,9])
L = max(len(a), len(b), len(c))
def pad(x, L): 
    out = np.zeros(L, dtype=int)
    out[:len(x)] = x
    return out
M = np.vstack([pad(a,L), pad(b,L), pad(c,L)])
print(M)

##### 17) Selección top-k por columna
Enunciado:
Crea A 6×4 con reales N(0,1). Para cada columna, obtén los índices de los 2 valores más grandes y construye un array booleano B 6×4 con True en esas posiciones y False en el resto.



```python
#Solución:

np.random.seed(7)
A = np.random.randn(6,4)
idx = np.argsort(A, axis=0)[-2:]  # filas de top-2 por columna
B = np.zeros_like(A, dtype=bool)
for col in range(A.shape[1]):
    B[idx[:, col], col] = True
print(A, "\n---\n", B)
```

##### 18) Distancias fila-fila (L2) de una matriz
Enunciado:
Dada X de 5×3, calcula la matriz de distancias euclídeas entre todas las filas (usando solo NumPy y broadcasting).

La distancia euclídea:||x_i - x_j||^2 = ||x_i||^2 + ||x_j||^2 - 2 x_i·x_j


```python
# 18 Solución:
np.random.seed(8)
X = np.random.randn(5,3)
# ||x_i - x_j||^2 = ||x_i||^2 + ||x_j||^2 - 2 x_i·x_j
sq_norms = (X**2).sum(axis=1, keepdims=True)
D2 = sq_norms + sq_norms.T - 2*(X @ X.T)
D = np.sqrt(np.maximum(D2, 0))
print(D)
```
