# 04 - Control de Flujo y Funciones — **Soluciones**

```

# Condicionales
x = 10
if x > 0:
    resultado = "positivo"
elif x == 0:
    resultado = "cero"
else:
    resultado = "negativo"
resultado
```

```

# Bucles
suma = 0
for i in range(1, 6):
    suma += i
suma
```

```

# Funciones
def cuadrado(n):
    return n ** 2

valores = [1, 2, 3, 4]
[cuadrado(v) for v in valores]
```
