### Operaciones con listas

A continuación, se muestran ejemplos de las principales operaciones que se pueden realizar con listas en Python:

```python
# Añadir elementos
numeros = [1, 2, 3]
numeros.append(4)  # Añade al final
print("Después de append(4):", numeros)

numeros.extend([5, 6])  # Añade múltiples elementos
print("Después de extend([5, 6]):", numeros)

numeros.insert(0, 0)  # Inserta en posición 0
print("Después de insert(0, 0):", numeros)
```

```python
# Eliminar elementos
letras = ['a', 'b', 'c', 'b', 'd']
letras.remove('b')  # Elimina primera ocurrencia de 'b'
print("Después de remove('b'):", letras)

eliminado = letras.pop()  # Elimina y devuelve el último elemento
print("Después de pop():", letras, "- Elemento eliminado:", eliminado)

eliminado2 = letras.pop(1)  # Elimina y devuelve elemento en índice 1
print("Después de pop(1):", letras, "- Elemento eliminado:", eliminado2)

letras.clear()  # Elimina todos los elementos
print("Después de clear():", letras)
```

```python
# Buscar y contar
animales = ['gato', 'perro', 'gato', 'pájaro', 'gato']
indice = animales.index('perro')  # Índice de primera ocurrencia
print("Índice de 'perro':", indice)

conteo = animales.count('gato')  # Número de ocurrencias
print("Número de 'gato':", conteo)

longitud = len(animales)  # Longitud de la lista
print("Longitud de la lista:", longitud)

# Ordenar y invertir
numeros_desordenados = [3, 1, 4, 1, 5]
numeros_desordenados.sort()  # Ordena ascendentemente
print("Después de sort():", numeros_desordenados)

numeros_desordenados.reverse()  # Invierte el orden
print("Después de reverse():", numeros_desordenados)
```

```python
# Slicing y otras operaciones
numeros = [10, 20, 30, 40, 50]
sub_lista = numeros[1:4]  # Elementos del índice 1 al 3
print("Slicing [1:4]:", sub_lista)

sub_lista2 = numeros[:3]  # Primeros 3 elementos
print("Slicing [:3]:", sub_lista2)

sub_lista3 = numeros[::2]  # Cada 2 elementos
print("Slicing [::2]:", sub_lista3)

# Operaciones con elementos
print("¿30 está en la lista?", 30 in numeros)
print("Mínimo:", min(numeros))
print("Máximo:", max(numeros))
print("Suma:", sum(numeros))

# Copiar lista
copia = numeros.copy()
print("Copia de la lista:", copia)
```

```python
# Listas por comprensión
cuadrados = [x**2 for x in range(1, 6)]
print("Cuadrados del 1 al 5:", cuadrados)

pares = [x for x in range(10) if x % 2 == 0]
print("Números pares del 0 al 9:", pares)

palabras = ["hola", "mundo", "python"]
mayusculas = [palabra.upper() for palabra in palabras]
print("Palabras en mayúsculas:", mayusculas)
```

## ✅ Autotests 04

```python
def check(cond, ok, fail):
    if cond: print("✔", ok)
    else: raise AssertionError("✘ " + fail)

check(frutas[0]=="manzana" and "kiwi" in frutas, "Lista 'frutas' correcta.", "Error en lista 'frutas'.")
check(coordenadas==(10,20) and isinstance(coordenadas, tuple), "Tupla 'coordenadas' correcta.", "Error en tupla.")
check(persona.get("edad",0) == 26, "Diccionario actualizado correctamente.", "Edad no actualizada.")
check("amarillo" in colores and len(colores)>=4, "Conjunto 'colores' correcto.", "Error en conjunto.")
```
