## Ejercicio 1 — `f(x)=x^3+2x` y grad en x=3

```
try:
    import jax, jax.numpy as jnp
    def f(x):
        return x**3 + 2*x
    df = jax.grad(f)
    deriv_3 = df(3.0)
except Exception as e:
    deriv_3 = None
    print("JAX no disponible:", e)
deriv_3
```

## Ejercicio 2 — `jit` en g(a,b)=a*b + a**2

```
try:
    import time, numpy as np
    import jax, jax.numpy as jnp

    @jax.jit
    def g(a, b):
        return a*b + a**2

    a_np = np.random.rand(200_000)
    b_np = np.random.rand(200_000)
    a_j = jnp.array(a_np); b_j = jnp.array(b_np)

    t0 = time.time()
    out1 = g(a_j, b_j).block_until_ready()
    t1 = time.time()
    out2 = g(a_j, b_j).block_until_ready()
    t2 = time.time()

    times = (t1-t0, t2-t1)
except Exception as e:
    times = None
    print("JAX no disponible:", e)
times
```

## Ejercicio 3 — `vmap` aplicar h(x)=x^2+1 a 10 elementos

```
try:
    import jax, jax.numpy as jnp
    def h(x):
        return x**2 + 1
    hv = jax.vmap(h)
    xs = jnp.arange(10.0)
    ys = hv(xs)
except Exception as e:
    ys = None
    print("JAX no disponible:", e)
ys
```

## ✅ Autotests 08

```
def check(cond, msg_ok, msg_fail):
    if cond:
        print("✔", msg_ok)
    else:
        raise AssertionError("✘ " + msg_fail)

# Test 1: derivada en x=3
if deriv_3 is not None:
    check(abs(float(deriv_3) - (3*3**2 + 2)) < 1e-6, "grad f(3) correcto.", "grad f(3) incorrecto.")
else:
    print("ℹ JAX no disponible: se omite el test de grad.")

# Test 2: tiempos jit
if times is not None:
    check(len(times)==2 and all(t>=0 for t in times), "Tiempos medidos para jit.", "No se midieron tiempos correctamente.")
else:
    print("ℹ JAX no disponible: se omite el test de jit.")

# Test 3: vmap salida
if ys is not None:
    import numpy as np
    check(np.allclose(np.array(ys), np.arange(10.0)**2 + 1), "vmap(h) aplicado correctamente.", "Resultado de vmap incorrecto.")
else:
    print("ℹ JAX no disponible: se omite el test de vmap.")
```
