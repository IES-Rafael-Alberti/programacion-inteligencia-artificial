## Bloque B — Filtro y columna derivada

1. Filtra los empleados con `salario > 3000` y muestra `nombre` y `salario`.
2. Crea `salario_anual = salario * 12`.


```r
print(empleados[empleados$salario > 3000, c("nombre", "salario")])
empleados$salario_anual <- empleados$salario * 12
print(empleados)
```

## Bloque C — Resumen por grupo

1. Calcula la media de `salario` por `nivel` con `aggregate(salario ~ nivel, data = empleados, FUN = mean)`.


```r
print(aggregate(salario ~ nivel, data = empleados, FUN = mean))
```
