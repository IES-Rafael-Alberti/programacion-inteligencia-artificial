# 03 - Operadores — **Soluciones**

Aritméticos, comparación, lógicos y asignación.

```

a, b = 5, 3
arit = (a+b, a-b, a*b, a/b, a//b, a % b, a**b)
comp = (a == b, a != b, a > b, a < b, a >= b, a <= b)
logicos = (a > 0 and b > 0, a > 0 or b < 0, not(a > 0))

a += 2
asig = a

arit, comp, logicos, asig
```
