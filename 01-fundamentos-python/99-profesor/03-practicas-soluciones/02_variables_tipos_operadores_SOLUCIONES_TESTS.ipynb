## Ejercicio A — Crea variables y comprueba tipos

### ✅ Solución

```
nombre = "Ana"
edad = 21
altura = 1.68
es_estudiante = True

tipos = (type(nombre), type(edad), type(altura), type(es_estudiante))
tipos
```

## Ejercicio B — Operadores aritméticos y lógicos

### ✅ Solución

```
a, b = 10, 3
ops = {
    "suma": a+b,
    "resta": a-b,
    "mul": a*b,
    "div": a/b,
    "floordiv": a//b,
    "mod": a % b,
    "pow": a ** b,
    "gt": a > b,
    "eq": a == b,
    "and": (True and False),
    "or":  (True or False),
    "not": (not True)
}
ops
```

---

## ✅ Autotests 02

```
def check(cond, ok, fail):
    if cond: print("✔", ok)
    else: raise AssertionError("✘ " + fail)

check(isinstance(nombre, str) and isinstance(edad, int) and isinstance(altura, float) and isinstance(es_estudiante, bool),
      "Tipos correctos para nombre/edad/altura/es_estudiante.",
      "Algún tipo no coincide.")
check(ops["suma"] == 13 and ops["floordiv"] == 3 and ops["mod"] == 1 and ops["pow"] == 1000,
      "Operadores aritméticos correctos.", "Resultados aritméticos incorrectos.")
check(ops["gt"] and not ops["eq"], "Operadores relacionales coherentes.", "Relacionales incoherentes.")
check(ops["and"] == False and ops["or"] == True and ops["not"] == False,
      "Operadores lógicos correctos.", "Lógicos incorrectos.")
```
