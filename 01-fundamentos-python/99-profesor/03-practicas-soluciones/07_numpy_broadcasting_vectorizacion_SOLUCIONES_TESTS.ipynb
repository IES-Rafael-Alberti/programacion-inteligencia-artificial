## Ejercicio 1 — Suma con broadcasting

```
import numpy as np

X = np.array([[1,2,3],[4,5,6]])   # (2,3)
v = np.array([10,20,30])          # (3,)
Y = X + v
X.shape, v.shape, Y.shape
```

## Ejercicio 2 — Comparar tiempos bucle vs vectorizado

```
import time
n = 300_000
a = np.random.rand(n)
b = np.random.rand(n)

t0 = time.time()
out_loop = np.empty_like(a)
for i in range(n):
    out_loop[i] = a[i]*b[i] + a[i]
t1 = time.time()

t2 = time.time()
out_vec = a*b + a
t3 = time.time()

(t1-t0, t3-t2, np.allclose(out_loop, out_vec))
```

## Ejercicio 3 — Sistema 3x3 con linalg.solve

```
M = np.array([[2., -1., 0.],
              [ -1., 2., -1.],
              [0., -1., 2.]])
b = np.array([1., 0., 1.])

x = np.linalg.solve(M, b)
residuo = M @ x - b
x, residuo, np.allclose(M @ x, b)
```
