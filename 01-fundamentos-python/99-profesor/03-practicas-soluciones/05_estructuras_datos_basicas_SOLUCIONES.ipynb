# 05 - Estructuras de Datos Básicas — **Soluciones**

```

# Listas
numeros = [1, 2, 3]
numeros.append(4)
numeros[1] = 20
numeros
```

```

# Tuplas
tupla = (1, 2, 3)
len(tupla)
```

```

# Diccionarios
alumno = {"nombre": "Ana", "edad": 21}
alumno["curso"] = "IA"
alumno
```

```

# Conjuntos
conjunto = {1, 2, 2, 3}
conjunto.add(4)
conjunto
```
